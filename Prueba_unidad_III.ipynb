# Ejercicio 1: Control de Tren

### Descripción
Simula el sistema de control de un tren que debe responder a señales en la vía. Las señales pueden ser **"verde"**, **"amarillo"** o **"rojo"**. El tren debe ajustar su velocidad base según la señal actual y la distancia hasta la próxima estación: 

- **Señal verde**: 
  - Acelera a **100 km/h** si la distancia es mayor a 5 km. 
  - Mantiene **80 km/h** en caso contrario. 
- **Señal amarilla**: 
  - Reduce la velocidad a **50 km/h** si la distancia es mayor a 3 km. 
  - De lo contrario, reduce a **30 km/h**. 
- **Señal roja**: 
  - Frena hasta **0 km/h** inmediatamente. 
- **Obstáculo en la vía**: 
  - El tren debe frenar a **0 km/h**, independientemente de la señal. 

--- 
### ✅ Salida esperada 
El programa debe devolver la velocidad actual del tren en base a las condiciones dadas. 

**Ejemplo:** 
- Entrada: `señal="amarillo", distancia=4, obstáculo=False`
- Salida: `Velocidad: 50 km/h`

### Restricciones

- Debe utilizar estructuras condicionales.
- No se permite crear listas ni funciones.

````python
#Entrada de datos

senal = input("Ingrese el color de la señal: ").lower()
distancia = int(input("Ingrese la distancia: "))
obstaculo = input("Hay obstaculo? (Ingrese si es True/False)").lower() == "true"
velocidad = 0

# Procesado de datos
if obstaculo == True:
    velocidad=0
else: 
    if senal == "verde" and  distancia > 5:
        velocidad = 100
        print ("Acelera 100 km/h")
    elif senal== "verde":
        velocidad = 80
        print ("Mantiene 80km/h")
    elif senal == "amarrillo" and distancia > 3:
        velocidad = 50
        print ("Mantiene a 50km/h")
    elif senal == "amarillo":
        velocidad = 30
        print ("Mantiene a 30km/h")
    elif senal == "rojo":
        velocidad = 0
        print ("Detenerse a 0 km/h")
print (f"La velocidad es: {velocidad} km/h")


````

# Ejercicio 2: Buscador del Próximo Año Bisiesto

### Descripción
Una empresa de planificación necesita conocer cuál será el próximo año bisiesto después de un año dado.

Un año es bisiesto cuando:

- Es divisible entre 4 y no entre 100.
- O es divisible entre 400.

El programa debe solicitar un año y utilizar un ciclo para encontrar el siguiente año bisiesto.

### ✅ Salida esperada

**Ejemplo:**
- Entrada: `2025`  
- Salida: `El próximo año bisiesto es: 2028`
____________

- Entrada: `2096`  
- Salida: `El próximo año bisiesto es: 2104`

### Restricciones

- Debe utilizar un ciclo **while** para encontrar el próximo año bisiesto.
- Debe utilizar estructuras condicionales.
- No se permite ingresar varios años.

````python
anio_actual = int(input("Ingrese el año actual: "))
anio_bisiesto = anio_actual + 1

while True:
    if anio_bisiesto % 400 == 0 or (anio_bisiesto % 4 == 0 and anio_bisiesto % 100 != 0):
        break
    else: 
        anio_bisiesto += 1
print (f"Elsiguiente año bisiesto es: {anio_bisiesto}")

````

# Ejercicio 3: Secuencia Misteriosa

### Descripción
Un laboratorio matemático está estudiando una secuencia numérica especial.

Dado un número entero positivo:

- Si el número es par, se divide enteramente entre 2.
- Si el número es impar, se multiplica por 3 y se suma 1.

El proceso se repite hasta que el número llegue a 1.

### ✅ Salida esperada
El programa debe mostrar cada valor generado y contar cuántos pasos fueron necesarios.

**Ejemplo:**
- Entrada: `12`  
- Salida: `Secuencia: 12, 6, 3, 10, 5, 16, 8, 4, 2, 1; Pasos: 9`
___

- Entrada: `45`  
- Salida: `Secuencia: 45, 136, 68, 34, 17, 52, 26, 13, 40, 20, 10, 5, 16, 8, 4, 2, 1; Pasos: 16`

### Restricciones

- Debe utilizar un ciclo while.
- Debe utilizar estructuras condicionales.
- El número ingresado debe ser mayor que cero.

````python
numero = int(input("Ingrese un numero entero positivo: "))
pasos = 0
secuencia = ""

if numero <= 0:
    print("El numero ingresado es invalido, por favor ingrese su numero nuevamente ")
else:
    
    while numero != 1:
        if numero % 2 == 0:
            numero = numero // 2
        else:
            numero = numero * 3 + 1
        pasos += 1
        secuencia = secuencia + str(numero) + ", "
print(f"Secuencia: {secuencia} ")
print(f"El numero de pasos necesarios es: {pasos}")

````

# Ejercicio 4: Centro de Control de Robots

### Descripción
Un centro de automatización controla varios robots industriales.

- El usuario indicará cuántos robots desea procesar.
- Para cada robot se solicitará su estado actual:

    - `"activo"`
    - `"espera"`
    - `"mantenimiento"`
    - `"apagado"`

El programa debe utilizar la **estructura match** para determinar la acción correspondiente:

| Estado        | Acción                  |
|---------------|-------------------------|
| Activo        | Operando normalmente    |
| Espera        | Esperando instrucciones |
| Mantenimiento | Revisar componentes     |
| Apagado       | Sistema detenido        |

Además:
- Debe llevar un conteo de cuántos robots se encuentran en cada estado.
- Al finalizar, mostrar un resumen general.
---

### ✅ Salida esperada

El programa debe mostrar la acción correspondiente para cada robot procesado y, al finalizar, presentar un resumen con la cantidad de robots en cada estado.

**Entrada**

```text
Cantidad de robots: 4

activo
espera
activo
apagado
```
___
**Salida**

```text
Robot 1: Operando normalmente
Robot 2: Esperando instrucciones
Robot 3: Operando normalmente
Robot 4: Sistema detenido

Resumen:
Activo: 2
Espera: 1
Mantenimiento: 0
Apagado: 1
```
___
### Restricciones

- Debe utilizar un ciclo for.
- Debe utilizar la estructura match.
- No se permite utilizar listas.

````python
# Aquí se resuelve el ejercicio

numero_robot = int(input("Ingrese la cantidad de robot (enteros positivos): "))
r_activo = 0
r_espera = 0
r_mantenimiento = 0
r_apagado = 0
estado = ""


if numero_robot <= 0:
    print ("El numero de robots ingresado es invalido, por favor ingrese su numero de nuevo") 


print (f"Ingrese 1 si el robot esta activo\nIngrese 2 si el robot esta en espera\nIngrese 3 si el robot esta en mantenimiento\nIngrese 4 si el robot esta apagado\n")
for i in range (1, numero_robot + 1):
    estado_caso = input(f"Ingrese el estado del robot {i}: ")
    
    match estado_caso:
        case "1":
            print (f"El robot {i} opera con normalidad")
            r_activo += 1
            estado = estado + (f"\nRobot {i}: activo ")
        case "2":
            print (f"El robot {i} se encuntra en espera")
            r_espera += 1
            estado = estado + (f"\nRobot {i}: espera ")

        case "3":
            print(f"El robot {i} se encuentra en mantenimiento")
            r_mantenimiento += 1
            estado = estado +  (f"\nRobot {i}: mantenimiento ")
        case "4":
            print(f"El robot {i} se encuentra apagado")
            r_apagado += 1
            estado = estado + (f"\nRobot {i}: apagado ")

        case _:
            print ("Estado inválido, intenta de nuevo")
            break

print (estado)
print (f"\nResumen\nActivo: {r_activo}\nEspera: {r_espera}\nMantenimiento: {r_mantenimiento}\nApagado: {r_apagado}")
    


# como 2 horas 
````
